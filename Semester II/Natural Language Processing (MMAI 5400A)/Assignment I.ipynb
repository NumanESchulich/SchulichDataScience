Step 0: Install necessary packages

```python
import subprocess
import sys


def check_and_install_package(package_name):
    """
    Function to check if a package is installed. If the package is not installed, 
    it will attempt to install it using pip.
    
    Parameters:
        package_name (str): The name of the package to check and install.
    """
    try:
        __import__(package_name)
        print(f"'{package_name}' is already installed.")
    except ImportError:
        print(f"'{package_name}' is not installed. Installing...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", package_name])


# Check if 'requests' and 'beautifulsoup4' are installed; if not, install them.
check_and_install_package("requests")
check_and_install_package("bs4")

# Import necessary libraries after ensuring they're installed.
import requests
from bs4 import BeautifulSoup
```

Step 1: Download the HTML for the CyberPowerPC's Review URL

```python
def download_html(url):
    """
    Downloads the HTML content from the provided URL.

    Parameters:
        url (str): The URL of the webpage to download.

    Returns:
        str: The HTML content of the page, or None if an error occurs.
    """
    try:
        response = requests.get(url)
        response.raise_for_status()  # Check if the request was successful
        return response.text
    except requests.exceptions.RequestException as e:
        print(f"Error fetching the webpage: {e}")
        return None


if __name__ == "__main__":
    url = "https://ca.trustpilot.com/review/www.cyberpowerpc.com"
    html_content = download_html(url)
    if html_content:
        print("HTML content successfully downloaded.")
    else:
        print("Failed to download HTML content.")
```

Step 2: Extracting Total Number of Reviews

```python
import re
from bs4 import BeautifulSoup


def extract_total_reviews(html_content):
    """
    Extracts the total number of reviews from the HTML content by searching
    for 'reviewCount'.
    
    Parameters:
        html_content (str): The HTML content of the page.
        
    Returns:
        int: The total number of reviews if found, None otherwise.
    """
    # Search for the reviewCount pattern in the HTML
    match = re.search(r'"reviewCount":"(\d+)"', html_content)
    
    if match:
        try:
            # Extract the review count from the matched group
            review_count = int(match.group(1))
            return review_count
        except ValueError as e:
            print(f"Error extracting review count: {e}")
            return None
    else:
        print("Could not find 'reviewCount' in the page.")
        return None


if __name__ == "__main__":
    # Extract the total number of reviews (assuming html_content is loaded externally)
    total_reviews = extract_total_reviews(html_content)
    if total_reviews:
        print(f"Total number of reviews: {total_reviews}")
    else:
        print("Failed to extract the total number of reviews.")
```
