Logistic Regression Model

```python
import pandas as pd
import os
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report

# Set the file path
file_path = os.path.join(os.getcwd(), 'midterm_parttwo.csv')

# Load the dataset
data = pd.read_csv(file_path)

# Map categorical columns to numeric values for model compatibility
data['Requested Credit Amount'] = data['Requested Credit Amount'].map({'Low': 1, 'Medium': 2, 'High': 3})
data['Number of Dependents'] = data['Number of Dependents'].map({'No dependent': 0, 'Less than two': 1, 'More than two': 2})
data['Monthly Income'] = data['Monthly Income'].map({'Very Low': 1, 'Low': 2, 'Moderate': 3, 'High': 4, 'Very High': 5})
data['Monthly Expense'] = data['Monthly Expense'].map({'Very Low': 1, 'Low': 2, 'Moderate': 3, 'High': 4, 'Very High': 5})
data = pd.get_dummies(data, columns=['Marital Status'], drop_first=True)

# Separate the target variable before imputing missing values
y = data['Credit Rating'].map({'Positive': 1, 'Negative': 0})
X = data.drop(columns=['Credit Rating'])

# Impute missing values with the median for numeric columns in X
X = X.apply(lambda col: col.fillna(col.median()) if col.dtype in ['float64', 'int64'] else col)

# Split the data into training (50%) and test (50%) sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

# Initialize and fit a logistic regression model
model = LogisticRegression(max_iter=1000, solver='liblinear')
model.fit(X_train, y_train)

# Predict on the test set
y_pred = model.predict(X_test)

# Print the classification report and confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=["Negative", "Positive"], zero_division=0))
```

Model with Threshold for Top 15% Approval

```python
import pandas as pd
import os
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report

# Set the file path
file_path = os.path.join(os.getcwd(), 'midterm_parttwo.csv')

# Load the dataset
data = pd.read_csv(file_path)

# Map categorical columns to numeric values for model compatibility
data['Requested Credit Amount'] = data['Requested Credit Amount'].map({'Low': 1, 'Medium': 2, 'High': 3})
data['Number of Dependents'] = data['Number of Dependents'].map({'No dependent': 0, 'Less than two': 1, 'More than two': 2})
data['Monthly Income'] = data['Monthly Income'].map({'Very Low': 1, 'Low': 2, 'Moderate': 3, 'High': 4, 'Very High': 5})
data['Monthly Expense'] = data['Monthly Expense'].map({'Very Low': 1, 'Low': 2, 'Moderate': 3, 'High': 4, 'Very High': 5})
data = pd.get_dummies(data, columns=['Marital Status'], drop_first=True)

# Separate the target variable before imputing missing values
y = data['Credit Rating'].map({'Positive': 1, 'Negative': 0})
X = data.drop(columns=['Credit Rating'])

# Impute missing values with the median for numeric columns in X
X = X.apply(lambda col: col.fillna(col.median()) if col.dtype in ['float64', 'int64'] else col)

# Split the data into training (50%) and test (50%) sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

# Initialize and fit a logistic regression model
model = LogisticRegression(max_iter=1000, solver='liblinear')
model.fit(X_train, y_train)

# Predict probabilities on the test set
y_probs = model.predict_proba(X_test)[:, 1]  # Probabilities for the "Positive" class

# Calculate the threshold for top 15% approval
threshold = sorted(y_probs, reverse=True)[int(0.15 * len(y_probs)) - 1]

# Apply the new threshold to make predictions
y_pred = (y_probs >= threshold).astype(int)

# Print the new threshold value
print(f"New Threshold for 15% Approval Rate: {threshold:.4f}")

print("\nConfusion Matrix with 15% Approval Rate:")
print(confusion_matrix(y_test, y_pred))

# Print the classification report and confusion matrix
print("\nClassification Report with 15% Approval Rate:")
print(classification_report(y_test, y_pred, target_names=["Negative", "Positive"], zero_division=0))
```
